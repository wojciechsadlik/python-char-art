```python
from PIL import Image, ImageFont
import matplotlib.pyplot as plt
from modules.img_processing import preprocess_img, DITHER_MODES
from modules.palette_generator import get_asciis, add_ansi_256_colors_grayscale
from modules.ansi_colorizer import AnsiColorizer, reset_code
from modules.img2chars_mono_converter import Img2MonoCharsConverter
```

```python
IMG_PATH = "../imgs/peppers.tiff"
FONT_PATH = "../fonts/Inconsolata-Regular.ttf"

img = Image.open(IMG_PATH).convert("RGB")
font = ImageFont.truetype(FONT_PATH, 32)
```

```python
proc_img = preprocess_img(img=img, scale_factor=0.1, sharpness=2, grayscale=True)
plt.imshow(proc_img, cmap='gray', vmin=0, vmax=255, interpolation='none')
plt.show()

converter = Img2MonoCharsConverter(
    symbols=get_asciis(),
    font=font,
    general_mapping_palette_shape=(12,12)
)

ascii_arr = converter.convert(proc_img)
for i in range(len(ascii_arr)):
    for j in range(len(ascii_arr[i])):
        print(ascii_arr[i][j], sep='', end='')
    print('\n', end='')
```

```python
proc_img = preprocess_img(img=img, scale_factor=0.2, sharpness=2, grayscale=True)
plt.imshow(proc_img, cmap='gray', vmin=0, vmax=255, interpolation='none')
plt.show()

converter = Img2MonoCharsConverter(
    symbols=get_asciis(),
    font=font,
    general_mapping_palette_shape=(12, 12),
    detailed_mapping_win_shape=(4, 2)
)

ascii_arr = converter.convert(proc_img)
for i in range(len(ascii_arr)):
    for j in range(len(ascii_arr[i])):
        print(ascii_arr[i][j], sep='', end='')
    print('\n', end='')
```

```python
proc_img = preprocess_img(img=img, scale_factor=0.1, sharpness=2, grayscale=True)
plt.imshow(proc_img, cmap='gray', vmin=0, vmax=255, interpolation='none')
plt.show()

converter = Img2MonoCharsConverter(
    symbols=add_ansi_256_colors_grayscale(get_asciis(), 8),
    font=font,
    general_mapping_palette_shape=(16,16),
    normalize_palette=False
)

ascii_arr = converter.convert(proc_img)
for i in range(len(ascii_arr)):
    for j in range(len(ascii_arr[i])):
        print(ascii_arr[i][j], sep='', end='')
    print('\n', end='')
```

```python
proc_img = preprocess_img(img=img, scale_factor=0.2, sharpness=2, grayscale=False)
plt.imshow(proc_img, cmap='gray', vmin=0, vmax=255, interpolation='none')
plt.show()

ansi_colorizer = AnsiColorizer(
    colored_fg=True,
    colored_bg=True,
    fg_brightness_scale=1.0,
    bg_brightness_scale=0.4,
    use_ansi_256_colors=False
)
converter = Img2MonoCharsConverter(
    symbols=get_asciis(),
    font=font,
    general_mapping_palette_shape=(12, 12),
    detailed_mapping_win_shape=(4, 2),
    ansi_colorizer=ansi_colorizer
)
ascii_arr = converter.convert(proc_img)
for i in range(len(ascii_arr)):
    for j in range(len(ascii_arr[i])):
        print(ascii_arr[i][j], sep='', end='')
    print(reset_code() + '\n', end='')
```
