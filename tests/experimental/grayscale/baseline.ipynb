```python
import sys
sys.path.append('../../../src')
sys.path.append('../../..')

import os
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
from image.processing import preprocess_img

from tests.experimental.grayscale.test_loop import similarity

IMG_SET_PATH = "../../../imgs/BSDS500"
IMG_PATHS = [os.path.join(IMG_SET_PATH, img_name)
             for img_name in os.listdir(IMG_SET_PATH)[:10]]
IMGS = [Image.open(img_path).convert("L") for img_path in IMG_PATHS]
print(len(IMGS))
display(IMGS[0])
```

```python
widths = []
heights = []
for img in IMGS:
    widths.append(img.size[0])
    heights.append(img.size[1])
print(min(widths), np.mean(widths), max(widths))
print(min(heights), np.mean(heights), max(heights))
```

```python
# scale

scale_factor_vals = np.linspace(1.0, 0.05, 5, endpoint=True)

results = []
for scale_factor in scale_factor_vals:
    similarities = []
    for img in IMGS:
        similarities.append(
            similarity(img,
                       preprocess_img(img, scale_factor=scale_factor))
        )
    results.append(similarities)

plt.boxplot(results,
            labels=np.round(scale_factor_vals, decimals=2))
plt.show()
```

```python
# edge enhancement

enhance_edges_vals = np.linspace(0.0, 2.0, 5, endpoint=True)

results = []
for enhance_edges in enhance_edges_vals:
    similarities = []
    for img in IMGS:
        similarities.append(
            similarity(img,
                       preprocess_img(img, enhance_edges=enhance_edges))
        )
    results.append(similarities)

plt.boxplot(results,
            labels=enhance_edges_vals)
plt.show()
```

```python
# contrast

contrast_vals = np.linspace(0.0, 2.0, 5, endpoint=True)

results = []
for contrast in contrast_vals:
    similarities = []
    for img in IMGS:
        similarities.append(
            similarity(img,
                       preprocess_img(img, contrast=contrast))
        )
    results.append(similarities)

plt.boxplot(results,
            labels=contrast_vals)
plt.show()
```

```python
# eq

eq_vals = np.linspace(0.0, 1.0, 5, endpoint=True)

results = []
for eq in eq_vals:
    similarities = []
    for img in IMGS:
        similarities.append(
            similarity(img,
                       preprocess_img(img, eq=eq))
        )
    results.append(similarities)

plt.boxplot(results,
            labels=eq_vals)
plt.show()
```
